```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(
    'Kelas D_Happiness and Satisfaction.csv',
    sep=';',
    decimal=','
)

df_happy = df[df['Happy Score (0-10)'] > 7.0].copy()

df_happy['Memenuhi'] = df_happy['GDP'] < 1.0

fig, ax = plt.subplots(figsize=(10, 6))

sns.barplot(
    data=df_happy,
    x='Country',
    y='GDP',
    hue='Memenuhi',
    ax=ax
)

ax.set_title('Negara dengan Happy Score > 7.0 dan GDP < 1.0')
ax.set_xlabel('Country')
ax.set_ylabel('GDP')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
