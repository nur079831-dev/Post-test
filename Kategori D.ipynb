```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
df = pd.read_csv(
    "Kelas D_Happiness and Satisfaction.csv",
    sep=";",
    decimal=","
)


df.head()
```

```python
fig, ax = plt.subplots(figsize=(10,6))


ax.hist(
    df["avg_income (&)"],
    bins=15,
    edgecolor="black"
)

ax.set_title(
    "Distribusi Pendapatan Rata-rata Global",
    fontsize=14,
    fontweight="bold"
)

ax.set_xlabel("Pendapatan Rata-rata (avg_income)")
ax.set_ylabel("Frekuensi Negara")

plt.tight_layout()
plt.show()
```
